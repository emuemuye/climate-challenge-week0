```python
from scipy import stats
```

```python
cols = ["T2M", "T2M_MAX", "T2M_MIN", "PRECTOTCORR", "RH2M", "WS2M", "WS2M_MAX"]
```

```python
z_scores = np.abs(stats.zscore(df[cols], nan_policy='omit'))
```

```python
outliers = (z_scores > 3).sum()
outliers
```

Outliers were identified using Z-score > 3. Depending on analysis needs, they may represent extreme climate events or measurement noise. For this analysis, we retain values to preserve climate variability patterns.

```python
import pandas as pd

ethiopia = pd.read_csv("../data/ethiopia_clean.csv")
kenya = pd.read_csv("../data/kenya_clean.csv")
sudan = pd.read_csv("../data/sudan_clean.csv")
tanzania = pd.read_csv("../data/tanzania_clean.csv")
nigeria = pd.read_csv("../data/nigeria_clean.csv")
```

```python
df_all = pd.concat([ethiopia, kenya, sudan, tanzania, nigeria])
df_all.head()
```

```python
import matplotlib.pyplot as plt

df_all.groupby(["date", "Country"])["T2M"].mean().unstack().plot(figsize=(12,6))
plt.title("Temperature Comparison Across Countries (2015–2026)")
plt.show()
```

```python
df_all.groupby("Country")["PRECTOTCORR"].mean().plot(kind="bar")
plt.title("Average Rainfall by Country")
plt.show()
```

```python
df_all.groupby("Country")[["T2M", "PRECTOTCORR"]].agg(["mean", "median", "std"])
```

```python
extreme_heat = df_all[df_all["T2M_MAX"] > 35].groupby("Country").size()
extreme_heat
```

```python
dry_days = df_all[df_all["PRECTOTCORR"] < 1].groupby("Country").size()
dry_days
```

```python
ranking = pd.DataFrame({
    "Extreme_Heat_Days": df_all[df_all["T2M_MAX"] > 35].groupby("Country").size(),
    "Dry_Days": df_all[df_all["PRECTOTCORR"] < 1].groupby("Country").size(),
    "Mean_Temperature": df_all.groupby("Country")["T2M"].mean()
}).fillna(0)

ranking
```

```python
ranking["Vulnerability_Score"] = (
    ranking["Extreme_Heat_Days"] +
    ranking["Dry_Days"] +
    ranking["Mean_Temperature"]
)

ranking.sort_values("Vulnerability_Score", ascending=False)
```

Climate Vulnerability Insights for COP32
-Countries with higher extreme heat days and dry days show significantly higher      climate stress exposure.
-Temperature trends indicate consistent warming across all countries, with           variation in intensity and variability.
-Precipitation patterns show strong seasonal instability in certain regions,         increasing drought and flood risk.
-Ethiopia shows a climate profile similar to its East African neighbors but with     notable seasonal rainfall variability.
-These findings support prioritizing adaptation finance for the most climate-        vulnerable countries to strengthen resilience and food security.
